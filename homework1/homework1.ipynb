## Задание 1

```python
import json

purchases = {}

with open('input_data/purchase_log.txt', encoding='utf-8') as f:
    for line in f:
        record = json.loads(line)
        user_id = record['user_id']
        if user_id == 'user_id':
            continue
        purchases[user_id] = record['category']

for user_id, category in list(purchases.items())[:10]:
    print(user_id, category)
```

## Задание 2

```python
import csv

visit_log_path = 'input_data/visit_log.csv'
funnel_path = 'output_data/funnel.csv'

with open(visit_log_path, encoding='utf-8', newline='') as infile, \
     open(funnel_path, 'w', encoding='utf-8', newline='') as outfile:

    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    header = next(reader)
    writer.writerow(header + ['category'])

    for user_id, source in reader:
        category = purchases.get(user_id)
        if category is not None:
            writer.writerow([user_id, source, category])
```
